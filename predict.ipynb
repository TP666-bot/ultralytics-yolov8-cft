```python
from pathlib import Path

from ultralytics import YOLO

# 与仓库内随附的示例图。Notebook 里没有 __file__，用 cwd（请在仓库根目录 ultralytics/ 下运行本 notebook）
try:
    _root = Path(__file__).resolve().parent
except NameError:
    _root = Path.cwd()
IMAGE = _root / "ultralytics" / "assets" / "bus.jpg"

model = YOLO("yolov8n.pt")

print(f"Running predict on {IMAGE}...", flush=True)
results = model.predict(
    source=IMAGE,
    conf=0.25,
    verbose=True,
    save=True,
)

```
